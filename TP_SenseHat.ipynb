## Mise en forme des données

Comme vous l'avez compris avec votre cours, une des limitations principales en IoT est la batterie. Afin de réduire au maximum l'utilisation de cette dernière, on va utiliser des standards et encoder les données.

Nous vous proposons dans cette partie d'utiliser le format CayenneLPP.  
https://docs.mydevices.com/docs/lorawan/cayenne-lpp

Cayenne Low Power Payload est conçu pour optimiser la payload contenant les données d'un ou plusieurs capteurs.  
Une trame CayenneLPP est composée d'un byte représentant le capteur, un byte représentant le type de la donnée et une N-bytes représentant la valeur :

| Data Channel | Data Type | Value | Data Channel | Data Type | Value | ... |
|-|-|-|-|-|-|-|
| 1byte | 1byte | N Bytes  |1byte | 1byte | M Bytes  | ...|

Nous vous proposons dans cette partie d'utiliser `pycayennelpp`(https://pypi.org/project/pycayennelpp/) afin d'empaqueter les données du capteur.

```python
try:
  from cayennelpp import LppFrame
except:
  !pip install pycayennelpp
  from cayennelpp import LppFrame
```
